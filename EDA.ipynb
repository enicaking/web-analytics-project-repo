## Formatting Salary

```python
# Converting monthly salary to yearly salary

from utilities import format_salary

# Find rows paid monthly
monthly = df["salary_unit"].str.contains("month", case=False, na=False)
# df.loc[monthly, ["salary_min", "salary_max"]].head(20)

# Checking for NaNs (exist in float64)
df.loc[monthly, ["salary_min", "salary_max"]].isna().any(axis=1).sum()
df.loc[monthly, ["salary_min", "salary_max"]].applymap(lambda x: (repr(x), type(x)))
```

```python
# Update salary with vectorial operation
df.loc[monthly, "salary_min"] = (df.loc[monthly, "salary_min"] * 12).round(2)
df.loc[monthly, "salary_max"] = (df.loc[monthly, "salary_max"] * 12).round(2)

# Update salary  
df.loc[monthly, "salary"] = (
    df.loc[monthly].apply(lambda row: format_salary(row["salary_min"], row["salary_max"], "yr"), axis=1)
)

# Update salary_unit
df.loc[monthly, "salary_unit"] = "year"
```

```python
valid = monthly & df["salary_min"].notna() & df["salary_max"].notna()
bad_rows = df.loc[monthly & ~valid, ["salary_min","salary_max","salary"]]
print(bad_rows)
len(bad_rows)
```

```python
# Converting hourly salary to yearly salary

# First, detect outliers that are already in yearly format

is_hourly = df["salary_unit"].str.contains("hour", case=False, na=False)
is_outlier = (is_hourly & (df["salary_min"]>1000))
is_part_time = df["job_type"].str.contains("Part-time", case=False, na=False)

# Find rows paid hourly, but filter out the part-time jobs and assume 20
hourly_outliers = (is_hourly & is_outlier)
hourly_rest = (is_hourly & ~is_outlier)
hourly_part = (is_hourly & is_part_time & ~is_outlier)
hourly_full = (is_hourly & ~is_part_time & ~is_outlier)
```

```python
# Directly handle outliers
# Update salary  
df.loc[hourly_outliers, "salary"] = (
    df.loc[hourly_outliers].apply(lambda row: format_salary(row["salary_min"], row["salary_max"], "yr"), axis=1)
)
# Update salary_unit
df.loc[hourly_outliers, "salary_unit"] = "year"
```

```python

# Checking for NaNs (exist in float64)
df.loc[hourly_part, ["salary_min", "salary_max"]].isna().any(axis=1).sum()
df.loc[hourly_part, ["salary_min", "salary_max"]].applymap(lambda x: (repr(x), type(x)))

```

```python
# Update salary with vectorial operation
df.loc[hourly_part, "salary_min"] = (df.loc[hourly_part, "salary_min"] * 20 * 12).round(2)
df.loc[hourly_part, "salary_max"] = (df.loc[hourly_part, "salary_min"] * 20 * 12).round(2)

df.loc[hourly_full, "salary_min"] = (df.loc[hourly_full, "salary_max"] * 40 * 12).round(2)
df.loc[hourly_full, "salary_max"] = (df.loc[hourly_full, "salary_max"] * 40 * 12).round(2)


# Update salary  
df.loc[hourly_rest, "salary"] = (
    df.loc[hourly_rest].apply(lambda row: format_salary(row["salary_min"], row["salary_max"], "yr"), axis=1)
)

# Update salary_unit
df.loc[hourly_rest, "salary_unit"] = "year"
```
